```python
def print_state(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def get_next_states(state):
    states = []
    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    # Order: Right, Down, Left, Up
    moves = [
        (0, 1),
        (1, 0),
        (0, -1),
        (-1, 0)
    ]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero = new_row * 3 + new_col

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            states.append(tuple(new_state))

    return states


def dfs(state, goal, path, visited, depth, max_depth):
    if state == goal:
        return path

    if depth >= max_depth:
        return None

    visited.add(state)

    for next_state in get_next_states(state):
        if next_state not in visited:
            result = dfs(
                next_state,
                goal,
                path + [next_state],
                visited,
                depth + 1,
                max_depth
            )

            if result is not None:
                return result

    return None


# Input initial state
print("Enter initial state (9 numbers, use 0 for blank):")
initial = tuple(map(int, input().split()))

# Input goal state
print("Enter goal state (9 numbers, use 0 for blank):")
goal = tuple(map(int, input().split()))

# Validate input
if (len(initial) != 9 or len(goal) != 9 or
        sorted(initial) != list(range(9)) or
        sorted(goal) != list(range(9))):
    print("Invalid input! Enter numbers 0 to 8 exactly once.")
else:
    max_depth = 30

    path = dfs(initial, goal, [initial], set(), 0, max_depth)

    if path:
        print("\nDFS Solution:")
        print("Number of moves:", len(path) - 1)
        print()

        for i, state in enumerate(path):
            print("Step", i)
            print_state(state)
    else:
        print("No solution found within the maximum depth.")
```

```python
def print_state(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def get_next_states(state):
    states = []
    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        (0, 1),
        (1, 0),
        (0, -1),
        (-1, 0)
    ]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero = new_row * 3 + new_col

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            states.append(tuple(new_state))

    return states


def depth_limited_dfs(state, goal, limit, path):
    if state == goal:
        return path

    if limit == 0:
        return None

    for next_state in get_next_states(state):
        if next_state not in path:
            result = depth_limited_dfs(
                next_state,
                goal,
                limit - 1,
                path + [next_state]
            )

            if result is not None:
                return result

    return None


def ids(initial, goal, max_depth=30):
    for depth in range(max_depth + 1):
        print("Searching at depth:", depth)

        result = depth_limited_dfs(
            initial, goal, depth, [initial]
        )

        if result is not None:
            return result

    return None


print("Enter initial state (9 numbers, use 0 for blank):")
initial = tuple(map(int, input().split()))

print("Enter goal state (9 numbers, use 0 for blank):")
goal = tuple(map(int, input().split()))

if (len(initial) != 9 or len(goal) != 9 or
        sorted(initial) != list(range(9)) or
        sorted(goal) != list(range(9))):
    print("Invalid input! Enter numbers 0 to 8 exactly once.")

else:
    path = ids(initial, goal)

    if path:
        print("\nIDS Solution:")
        print("Number of moves:", len(path) - 1)
        print()

        for i, state in enumerate(path):
            print("Step", i)
            print_state(state)
    else:
        print("No solution found within the maximum depth.")
```
