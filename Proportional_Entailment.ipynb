```python

from itertools import product

# Step 1: Define logical operators
def implies(a, b):
    return not a or b

# Step 2: Generate all possible models
models = list(product([True, False], repeat=3))

# Step 3: Evaluate Knowledge Base
def check_kb(P, Q, R):
    s1 = implies(Q, P)       # Q -> P
    s2 = implies(P, not Q)   # P -> ~Q
    s3 = Q or R              # Q V R

    return s1 and s2 and s3

# Step 4: Evaluate queries
def evaluate_queries(P, Q, R):
    return [
        R,                       # R
        implies(R, P),           # R -> P
        implies(Q, R),            # Q -> R
    ]

# Step 5: Print truth table
print("P Q R | Q->P P->~Q QVR | KB")
print("-" * 39)

valid_models = []

for P, Q, R in models:
    s1 = implies(Q, P)
    s2 = implies(P, not Q)
    s3 = Q or R

    kb = s1 and s2 and s3

    print(
        f"{int(P)} {int(Q)} {int(R)} |"
        f"  {int(s1)}    {int(s2)}    {int(s3)}  | {int(kb)}"
    )

    if kb:
        valid_models.append((P, Q, R))

# Step 6: Check entailment
queries = [
    ("R", lambda P, Q, R: R),
    ("R -> P", lambda P, Q, R: implies(R, P)),
    ("Q -> R", lambda P, Q, R: implies(Q, R))
]

print("\nModels in which KB is TRUE:")
for model in valid_models:
    print(tuple(int(x) for x in model))

print("\nEntailment Results:")

for name, query in queries:
    entails = all(query(*model) for model in valid_models)

    if entails:
        print("KB entails", name)
    else:
        print("KB does not entail", name)
```
