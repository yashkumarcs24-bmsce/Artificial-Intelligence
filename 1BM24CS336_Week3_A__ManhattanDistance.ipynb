```python
from heapq import heappush, heappop

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

def manhattan_distance(state):
    distance = 0

    for i in range(9):
        tile = state[i]

        if tile != 0:
            current_row, current_col = divmod(i, 3)
            goal_index = goal.index(tile)
            goal_row, goal_col = divmod(goal_index, 3)

            distance += abs(current_row - goal_row)
            distance += abs(current_col - goal_col)

    return distance

def get_neighbors(state):
    neighbors = []
    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        nr, nc = row + dr, col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_pos = nr * 3 + nc
            temp = list(state)
            temp[zero], temp[new_pos] = temp[new_pos], temp[zero]
            neighbors.append(tuple(temp))

    return neighbors

def astar(start):
    pq = []
    heappush(pq, (manhattan_distance(start), 0, start))

    parent = {start: None}
    cost = {start: 0}

    while pq:
        f, g, state = heappop(pq)

        if state == goal:
            path = []
            while state is not None:
                path.append(state)
                state = parent[state]
            return path[::-1]

        if g != cost[state]:
            continue

        for neighbor in get_neighbors(state):
            new_g = g + 1

            if neighbor not in cost or new_g < cost[neighbor]:
                cost[neighbor] = new_g
                parent[neighbor] = state
                new_f = new_g + manhattan_distance(neighbor)
                heappush(pq, (new_f, new_g, neighbor))

    return None

start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

path = astar(start)

if path:
    print("Solution found!")
    print("Number of moves:", len(path) - 1)

    for state in path:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
else:
    print("No solution exists.")
```
