```python
def is_variable(x):
    return isinstance(x, str) and x.startswith("?")


def occurs_check(var, expr, subst):
    expr = apply_substitution(expr, subst)

    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(occurs_check(var, arg, subst) for arg in expr[1:])

    return False


def apply_substitution(expr, subst):
    if is_variable(expr):
        if expr in subst:
            return apply_substitution(subst[expr], subst)
        return expr

    if isinstance(expr, tuple):
        return (expr[0],) + tuple(
            apply_substitution(arg, subst)
            for arg in expr[1:]
        )

    return expr


def unify(x, y, subst=None):
    if subst is None:
        subst = {}

    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    if x == y:
        return subst

    if is_variable(x):
        if occurs_check(x, y, subst):
            return None
        subst[x] = y
        return subst

    if is_variable(y):
        if occurs_check(y, x, subst):
            return None
        subst[y] = x
        return subst

    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0]:
            return None

        if len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst

    return None


# Example
expr1 = ("Eats", "?x", "Apple")
expr2 = ("Eats", "Riya", "?y")

result = unify(expr1, expr2)

if result is None:
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitution:", result)
    print("Unified Expression:", apply_substitution(expr1, result))
```
