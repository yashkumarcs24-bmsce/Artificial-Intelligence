```python
from itertools import combinations

def calculate_cost(state):
    cost = 0

    for i in range(4):
        for j in range(i + 1, 4):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def generate_neighbours(state):
    neighbours = []

    for i, j in combinations(range(4), 2):
        neighbour = state.copy()
        neighbour[i], neighbour[j] = neighbour[j], neighbour[i]
        neighbours.append((i, j, neighbour, calculate_cost(neighbour)))

    return neighbours


def hill_climbing(initial_state):
    current = initial_state.copy()
    step = 1

    while True:
        current_cost = calculate_cost(current)

        print(f"\nStep {step}")
        print("Current state:", current)
        print("Current cost:", current_cost)

        if current_cost == 0:
            print("Goal state reached!")
            return current

        neighbours = generate_neighbours(current)

        for i, j, state, cost in neighbours:
            print(f"Swap ({i}, {j}): {state}, Cost = {cost}")

        best = min(neighbours, key=lambda n: (n[3], n[0], n[1]))

        if best[3] >= current_cost:
            print("No better neighbour. Stopping.")
            return current

        current = best[2]
        step += 1


initial_state = [3, 1, 2, 0]
solution = hill_climbing(initial_state)

print("\nFinal solution:", solution)
print("Final cost:", calculate_cost(solution))
```
