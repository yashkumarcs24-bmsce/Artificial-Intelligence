```python
import random
import math


def calculate_cost(state):
    cost = 0

    for i in range(8):
        for j in range(i + 1, 8):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def generate_neighbour(state):
    neighbour = state.copy()

    i, j = random.sample(range(8), 2)
    neighbour[i], neighbour[j] = neighbour[j], neighbour[i]

    return neighbour


def simulated_annealing(initial_state):
    current = initial_state.copy()
    current_cost = calculate_cost(current)

    best = current.copy()
    best_cost = current_cost

    temperature = 100.0
    cooling_rate = 0.2

    min_temperature = 0.01

    step = 1

    while temperature > min_temperature:
        if best_cost == 0:
            break

        neighbour = generate_neighbour(current)
        neighbour_cost = calculate_cost(neighbour)

        delta = neighbour_cost - current_cost

        if delta < 0:
            current = neighbour
            current_cost = neighbour_cost
        else:
            probability = math.exp(-delta / temperature)

            if random.random() < probability:
                current = neighbour
                current_cost = neighbour_cost

        if current_cost < best_cost:
            best = current.copy()
            best_cost = current_cost

        print(f"Step {step}: Temperature = {temperature:.2f}, "
              f"Current Cost = {current_cost}, Best Cost = {best_cost}")

        temperature *= cooling_rate
        step += 1

    return best


initial_state = random.sample(range(8), 8)

print("Initial state:", initial_state)
print("Initial cost:", calculate_cost(initial_state))

solution = simulated_annealing(initial_state)

print("\nFinal solution:", solution)
print("Final cost:", calculate_cost(solution))

if calculate_cost(solution) == 0:
    print("Goal state reached!")
else:
    print("Solution not found.")
```

```python
import random
import math

N = 8


def calculate_cost(board):
    conflicts = 0

    for col1 in range(N):
        for col2 in range(col1 + 1, N):
            row1 = board[col1]
            row2 = board[col2]

            if (row1 == row2 or
                    abs(row1 - row2) == abs(col1 - col2)):
                conflicts += 1

    return conflicts


def get_neighbour(board):
    neighbour = board.copy()

    col = random.randrange(N)
    old_row = neighbour[col]

    new_row = random.randrange(N - 1)
    if new_row >= old_row:
        new_row += 1

    neighbour[col] = new_row
    return neighbour


def simulated_annealing():
    # Step 1: Fixed initial state (0-based rows)
    current = [0, 0, 0, 0, 0, 0, 0, 0]
    current_cost = calculate_cost(current)

    print("Initial state:", [r + 1 for r in current])
    print("Initial cost:", current_cost)

    best = current.copy()
    best_cost = current_cost

    # Step 2: Initial temperature
    T = 10.0
    cooling_rate = 0.99
    max_steps = 100000

    for step in range(1, max_steps + 1):

        # Stop if a solution is found
        if current_cost == 0:
            return current, current_cost, step - 1

        # Step 3: Generate a random neighbour
        next_state = get_neighbour(current)
        next_cost = calculate_cost(next_state)

        # Step 4: Calculate cost difference
        delta_E = current_cost - next_cost

        # Step 5: Accept or reject the neighbour
        if delta_E > 0:
            current = next_state
            current_cost = next_cost
        else:
            p = math.exp(delta_E / T)

            if random.random() < p:
                current = next_state
                current_cost = next_cost

        # Store the best state found
        if current_cost < best_cost:
            best = current.copy()
            best_cost = current_cost

        # Step 6: Decrease temperature
        T *= cooling_rate

        if T < 1e-8:
            break

    return best, best_cost, step


solution, cost, steps = simulated_annealing()

print("Best state:", [r + 1 for r in solution])
print("Best cost:", cost)
print("Iterations:", steps)

if cost == 0:
    print("Solution found!")
else:
    print("Best state found; solution not guaranteed.")
```
